#Word Frequency Counter
This program reads a text string and prints how many times each word appears in it.

**Plan**
1. Take the input text
2. Convert to lowercase
3. Remove punctuation
4. Split into words
5. Count each word
6. Sort and print the result

## Step 01: The input text
Store the sentence in a variable called 'text'.

```python
text = "The rain fell on the car. The rain fell on the ground."
print(text)
```

## Step 2: Convert to lowercase
"The" and "the" are same word, so we convert everything to lowercase.

```python
text = text.lower()
print(text)
```

## Step 3: Remove punctuation
Without this, "car." and "car" would be counted as different words.

```python
import string

for p in string.punctuation:
  text = text.replace(p, "")
  print(text)
```

## Step 4: Split into words
'split()' breaks the text into a list of words at every space.

```python
words = text.split()
print(words)
```

## Step 5: Count the words
We use a dictionary. The key is the word and the value is how many times it appeared.
If the word is already in the dictionary we add 1, otherwise we start it at 1.

```python
counts = {}

for word in words:
  if word in counts:
     counts[word] += 1

  else:
    counts[word]= 1
print(counts)
```

## Step 6: Sort and print
Sort by count, highest first, and print in the format 'word-count.'

```python
sorted_counts = sorted(counts.items(), key=lambda item: item[1], reverse=True)

for word, count in sorted_counts:
  print(f"{word}-{count}")
```

## Step 7: Put it all in a function
Reuse the program with any text.

```python
import string

def word_frequency(text):
  """Takes a string and prints the frequency of each word."""
  # lowercase and remove punctuation
  text = text.lower()
  for p in string.punctuation:
    text = text.replace(p, "")

  # split into words
  words = text.split()

  # count
  counts = {}
  for word in words:
    counts[word] = counts.get(word,0) + 1

  #sort and print
  sorted_counts = sorted(counts.items(), key=lambda item: item[1], reverse=True)
  result = ", ".join(f"{word}-{count}" for word, count in sorted_counts)
  print(result)

word_frequency("The rain fell on the car. The rain fell on the ground.")
```

## Step 8: Testing with other sentences

```python
word_frequency("Hello world! Python is a good programming language with latest features.")
word_frequency("To be, or not to be.")
```

Step 9: Using python's built-in Counter
Python has a built-in tool that does the counting for us.

```python
from collections import Counter

text = "The rain fell on the car. The rain fell on the ground."
words = text.lower().replace(".", "").split()
print(Counter(words).most_common())
```
